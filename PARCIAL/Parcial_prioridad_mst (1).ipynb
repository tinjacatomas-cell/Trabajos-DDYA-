### 5) Análisis de complejidad **[15 pts]**
* El programa utiliza dos algoritmos principales


        * Max - Heap
        * Kruskal

* Para Max - Heap tenemos que la complejidad es:

        *  O (log (n))

* Para Kruskal tenemos que la complejidad es:

        * Para calcular distancia = O(V^2)

        * Para ordenar las conexiones = O( E log (E) )

        * Para seleccionar las conexiones sin formar ciclos = O(E)


**Funcion T(n)**


* T(n) = O ( n + E Log(E) + V^2 )


**Mejor caso**

* Los elementos se organizan rapidamente y se realizan pocas operaciones. Se calculan las distancias y las conexiones ya estan ordenadas


* Complejidad del mejor caso

      * O(n^2)

**Peor Caso**

* Se deben ordenar todas las conexiones antes de aplicar el algoritmo de Kruskal

* Complejidad del peor caso

      * O( n^2 log(n) )
